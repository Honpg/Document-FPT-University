```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Step 1: Read data file into a Pandas DataFrame
file_path = 'Canada.csv'  # Replace with your file path
canada_data = pd.read_csv(file_path)

# Step 2: Process and manipulate the data

# Remove unnecessary columns
columns_to_keep = ['OdName', 'AreaName', 'RegName', 'DevName'] + [str(year) for year in range(1980, 2014)]
canada_data = canada_data[columns_to_keep]

# Rename columns for clarity
canada_data.rename(columns={'OdName': 'Country', 'AreaName': 'Continent', 'RegName': 'Region', 'DevName': 'DevelopmentStatus'}, inplace=True)

# Fill missing values with zeros
canada_data.fillna(0, inplace=True)
```

```python
canada_data.head()
```

```python
# Step 3: Generate line plots using Matplotlib

# Selecting countries for comparison
countries = ['India', 'China', 'Philippines']

# Plotting immigration trends for these countries
plt.figure(figsize=(10, 6))
for country in countries:
    country_data = canada_data[canada_data['Country'] == country].iloc[:, 4:].T
    country_data.columns = ['Immigrants']
    plt.plot(country_data, label=country)

plt.title('Immigration Trends to Canada: India vs. China vs. Philippines (1980-2013)')
plt.xlabel('Year')
plt.ylabel('Number of Immigrants')
plt.xticks(rotation=45)
plt.legend()
plt.grid(True)
plt.show()
```
