```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.backends.backend_pdf import PdfPages
from sklearn import svm
from sklearn.model_selection import train_test_split
from sklearn.svm import SVC
from sklearn.metrics import accuracy_score
```

```python

# Set seed for reproducibility
np.random.seed(3456)

# Generate random data
x = np.random.randn(50, 2)
y = np.repeat([-1, 1], [25, 25])
e = np.random.randn(25)
x[y == 1, :] = x[y == 1, :] + e.reshape(-1, 1)

# Create SVM model with linear kernel
mysvm = SVC(kernel='linear', C=10)

# Fit the SVM model
mysvm.fit(x, y)

# Predict using the trained SVM model
svm_pred = mysvm.predict(x)

# Calculate accuracy
accuracy = accuracy_score(y, svm_pred)
print(f'Accuracy: {accuracy * 100:.2f}%')

# Plot the data points
plt.scatter(x[:, 0], x[:, 1], c=y, cmap=plt.cm.Paired)

# Plot the decision boundary
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 30),
                     np.linspace(ylim[0], ylim[1], 30))
Z = mysvm.decision_function(np.c_[xx.ravel(), yy.ravel()])
Z = Z.reshape(xx.shape)

plt.contour(xx, yy, Z, colors='k', levels=[-1, 0, 1], alpha=0.5,
            linestyles=['--', '-', '--'])

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('SVM Decision Boundary with Linear Kernel')
plt.show()

x[y == 1, :] = x[y == 1, :] + e.reshape(-1, 1)

# Create SVM model with linear kernel
mysvm = SVC(kernel='linear', C=10)

# Fit the SVM model
mysvm.fit(x, y)

# Predict using the trained SVM model
svm_pred = mysvm.predict(x)

# Calculate accuracy
accuracy = accuracy_score(y, svm_pred)

print(f'Accuracy: {accuracy * 100:.2f}%')
```

```python
# Set seed for reproducibility
np.random.seed(3456)

# Generate random data
x = np.random.randn(50, 2)
y = np.repeat([-1, 1], [25, 25])
e = np.random.randn(25)
x[y == 1, :] = x[y == 1, :] + e.reshape(-1, 1)

# Define a custom tanh kernel function
def tanh_kernel(X, Y):
    return np.tanh(np.dot(X, Y.T))

# Create SVM model with precomputed tanh kernel
mysvm = SVC(kernel='precomputed', C=10)

# Compute the tanh kernel matrix
tanh_kernel_matrix = tanh_kernel(x, x)

# Fit the SVM model with the precomputed kernel matrix
mysvm.fit(tanh_kernel_matrix, y)

# Predict using the trained SVM model
svm_pred = mysvm.predict(tanh_kernel_matrix)

# Calculate accuracy
accuracy = accuracy_score(y, svm_pred)
print(f'Accuracy: {accuracy * 100:.2f}%')

# Plot the data points
plt.scatter(x[:, 0], x[:, 1], c=y, cmap=plt.cm.Paired)

# Plot the decision boundary
ax = plt.gca()
xlim = ax.get_xlim()
ylim = ax.get_ylim()

xx, yy = np.meshgrid(np.linspace(xlim[0], xlim[1], 30),
                     np.linspace(ylim[0], ylim[1], 30))
Z = mysvm.decision_function(tanh_kernel(np.c_[xx.ravel(), yy.ravel()], x))
Z = Z.reshape(xx.shape)

plt.contour(xx, yy, Z, colors='k', levels=[-1, 0, 1], alpha=0.5,
            linestyles=['--', '-', '--'])

plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.title('SVM Decision Boundary with Tanh Kernel')
plt.show()
```
