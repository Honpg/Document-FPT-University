```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# Load the CSV file
file_path = 'PhuLien_bochoi.csv'
data = pd.read_csv(file_path)

# Display the first few rows of the dataframe to understand its structure
data.head()
```

```python
# Set the style for the plots
sns.set(style="whitegrid")

# Prepare data for line plot
data_melted = data.melt(id_vars=['nam'], var_name='Thang', value_name='GiaTri')

# Create a figure for multiple plots
fig, axes = plt.subplots(3, 1, figsize=(15, 20))

# Line plot
sns.lineplot(x='nam', y='GiaTri', hue='Thang', data=data_melted, ax=axes[0], palette='viridis')
axes[0].set_title('Trend Over Years by Month')
axes[0].set_xlabel('Year')
axes[0].set_ylabel('Value')

# Heatmap
heatmap_data = data.set_index('nam')
sns.heatmap(heatmap_data.transpose(), cmap='coolwarm', ax=axes[1])
axes[1].set_title('Heatmap of Monthly Values Over Years')
axes[1].set_xlabel('Year')
axes[1].set_ylabel('Month')

# Box Plot
sns.boxplot(x='Thang', y='GiaTri', data=data_melted, ax=axes[2], palette='Set2')
axes[2].set_title('Monthly Value Distribution Across Years')
axes[2].set_xlabel('Month')
axes[2].set_ylabel('Value')

# Adjust layout
plt.tight_layout()

# Show the plots
plt.show()
```
