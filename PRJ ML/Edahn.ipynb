```python
import pandas as pd
import numpy as np

data_path = 'Impute_misvalues_hanoi.csv'
data = pd.read_csv(data_path)

data.head()
```

```python
data = data.loc[:, ~data.columns.str.contains('^Unnamed')]
data.drop('Average', axis=1, inplace=True)
data.head()
```

```python
data.drop_duplicates()
```

```python
data.isnull().sum()
```

```python
data['Date'] = pd.to_datetime(data['Date'], errors='coerce')
data['Date'].fillna(method='ffill', inplace=True)
```

```python
data.isnull().sum()
```

```python
data['Date'] = pd.to_datetime(data['Date'], errors='coerce')

# Format the 'Date' column to the desired format 'MM/DD/YYYY' without time information
data['Date'] = data['Date'].dt.strftime('%m/%d/%Y')
data.head()
```

```python
data.to_csv('Impute_misvalues_hanoi1.csv')
```
