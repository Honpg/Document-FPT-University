```python
import pandas as pd

data_path = 'Impute_misvalues_vuquang.csv'
data = pd.read_csv(data_path)

data.head()
```

```python
data.drop('X', axis=1, inplace=True)
```

```python
data.head()
```

```python
data.drop_duplicates()
```

```python
data.isnull().sum()
```

```python
data['Date'] = pd.to_datetime(data['Date'], errors='coerce')
data['Date'].fillna(method='ffill', inplace=True)
```

```python
data.head()
```

```python
data.isnull().sum()
```

```python
data['Date'] = pd.to_datetime(data['Date'], errors='coerce')

# Format the 'Date' column to the desired format 'MM/DD/YYYY' without time information
data['Date'] = data['Date'].dt.strftime('%m/%d/%Y')

# Display the first few rows to confirm the format
data.head()
```

```python
data.to_csv('Impute_misvalues_vuquang1.csv')
```
